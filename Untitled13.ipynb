```python
def my_funct(*kids):
    print("The youngest child is" + kids[2])

my_funct("Emil", "Tobias" "Linus")
```

```python
def my_funct2(*args):
    print("For type:", type(args))
    print("First arg:", args[0])
    print("Second arg:", args[1])
    print("Third arg:", args[2])
    print("All arg:", args)

my_funct2("Millicent", "Jemima", "Gold", "Precious", "Esther")
```

```python
def jem(greeting, *names):
    for name in names:
        print(greeting + " " + name)

jem("Hello", "Jemima", "Gold", "Precious", "Esther")
```

```python
def mill(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(mill(1, 2, 3, 4, 5))
print(mill(5, 10, 15, 20, 25, 30, 35, 40))
print(mill(6, 4))
```

```python
def new_num(*numbers):
    if len(numbers) == 0:
        return None
    max_num = numbers[0]
    for num in numbers:
        if num > max_num:
            max_num = num
    return max_num

print(new_num(3, 7, 2, 9, 14, 4))
```

```python
def new_key(**mrvar):
    print("Name:", mrvar["name"])
    print("Age:", mrvar["age"])
    print("All data:", mrvar)

new_key(name = "Millicent", age = 25, city = "Uyo")
```

```python
def nuw_fun(username, **detail):
    print("Username:", username)
    print("additional details:")
    for key, value in detail.items():
        print(" ", key + ":", value )

nuw_fun("milli123", age = 34, country = "Nigeria", hoby = "Teaching")
```
